# WeRateDogs Data Analysis Report


## Introduction

The WeRateDogs Twitter account is a popular social media presence that humorously rates dogs shared by users. This project involves data wrangling and analysis of tweets from the WeRateDogs account. The main goals were to clean and merge datasets, assess data quality and tidiness, and extract insights related to dog popularity, ratings, and social engagement.

Three key questions guided the analysis:
1. What types of dog breeds are rated the highest?
2. Does a higher dog rating correlate with more retweets or favorites?
3. Which stage of dogs (puppers, puppo, doggo, floofer) receives the most attention?

The cleaned dataset includes tweet-level information such as the rating numerator and denominator, number of retweets, favorites, predicted dog breeds, and the stage of the dog. The insights below are derived from this dataset.



## Insight 1: High Ratings Correlate With More Engagement

Tweets with higher ratings tend to receive more retweets and favorites. This suggests that users engage more with highly-rated dogs, possibly because these tweets are seen as more entertaining or positive. The plot below shows a clear upward trend between rating and engagement.



## Insight 2: Certain Dog Breeds Are Rated More Highly

Golden Retrievers, Labrador Retrievers, and Pembroke Welsh Corgis are among the top-rated breeds. These breeds often appear in highly-rated tweets, possibly due to their friendly appearance and popularity. The chart below shows average rating by breed for the top 10 most frequently occurring breeds.



## Insight 3: 'Pupper' Dogs Are the Most Common Stage

Among the four dog stages recorded (pupper, puppo, doggo, floofer), the 'pupper' stage is the most common. This may be due to the cuteness associated with younger dogs, making them more shareable and likely to be featured. The bar chart below visualizes the distribution of dog stages.

## Conclusion

The analysis of WeRateDogs tweets reveals interesting patterns in how dogs are rated and how those ratings affect social engagement. Higher ratings correlate with more retweets and favorites, certain breeds are more commonly rated highly, and 'pupper' is the most frequently seen stage. These insights not only reflect public preferences but can also inform content strategies for similar social media accounts.

This project showcases the importance of thorough data wrangling and the power of visualizations to support insights in social media data.
