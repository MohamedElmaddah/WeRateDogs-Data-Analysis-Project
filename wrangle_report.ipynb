# Wrangle Report

## Data Gathering

The data gathering process began smoothly with the import of the `twitter-archive-enhanced.csv` file. However, challenges arose during the import of the other two datasets. The `image_predictions.tsv` file was relatively straightforward once the appropriate `sep='\t'` argument was used.

The most significant challenge came with the `tweet_json.txt` file, originally expected to be obtained using the Twitter API. Setting up a developer account for access proved time-consuming. Fortunately, Udacity provided a pre-downloaded version of the data. Even then, parsing the file required using the JSON module to read it line by line into a usable format. Ultimately, all three datasets were successfully loaded into pandas DataFrames.

## Assessing Data

The assessment phase was essential in understanding the structure and content of the data. Initial exploration revealed numerous missing values, especially in columns like `in_reply_to_status_id`, `retweeted_status_id`, and some rating-related fields. Rather than dropping all missing data, I considered the context and relevance of each column, which helped guide cleaning decisions.

Both visual and programmatic assessments helped identify quality and tidiness issues. For example, some ratings were incorrectly parsed (e.g., numerators > 10), and dog stages (like 'pupper' and 'doggo') were scattered across multiple columns instead of being consolidated.

## Data Cleaning

Cleaning involved several steps. First, I addressed quality issues such as incorrect data types, duplicate tweets, and inconsistent formatting in rating scores. Then I tackled tidiness issues by reshaping the dog stages into a single column and merging the three datasets into one master DataFrame.

Throughout the process, I used pandas for operations like merging, filtering, and reshaping, and maintained copies of the data to avoid accidental loss. This structured, methodical approach ensured the final dataset was both clean and analysis-ready.
